```python
# How mny each classes are 
df["target"].value_counts()
```

```python
# plotting it for better visualization.
ax = df["target"].value_counts().plot(kind='bar', color=["salmon", "lightblue"]);

ax.set(xlabel="Class labels", 
       ylabel="total",
       title='Class weight')

legend_handles = [
    Patch(facecolor="salmon", label="Heart Disease Patient"),
    Patch(facecolor="lightblue", label="Not Heart Disease")
]

ax.legend(handles=legend_handles);

# let's save it to the local directory 
plt.savefig("../images/eda/class_weight_graph.png")
```

```python
# compare the target column with the `sex` column 
# based on our current data, we can see that if the women/female came there is 75% of chances to being the heart disesase patient.
pd.crosstab(df.target, df.sex)
```

```python
# visualize using matplotib 
ax = pd.crosstab(df.target, df.sex).plot(
    kind='bar',
    color=["salmon", "lightblue"],
    figsize=(10,6)
)

# setting the lagend handler
legend_handles=[
    Patch(facecolor="salmon", label="Male"),
    Patch(facecolor="lightblue", label="Female"),
]

# customizing the plot 
ax.set(xlabel="Target",
      ylabel="total",
      title="gender based target analysis")

# setting up legends 
ax.legend(handles=legend_handles);

# saving the figure 
plt.savefig("../images/eda/gender_based_heart_disease_analysis.png")
```
